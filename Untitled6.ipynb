```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import (
    StandardScaler,
    MinMaxScaler,
    LabelEncoder,
    OrdinalEncoder
)

pd.set_option("display.max_columns", None)
```

```python

data = pd.DataFrame({
    "Item": [
        "Coffee", "Latte", "Tea", "Orange Juice", "Sandwich",
        "Cake", "Espresso", "Cappuccino", "Iced Coffee", "Water",
        "Croissant", "Muffin", "Pizza Slice", "Smoothie",
        "Hot Chocolate", "Lemonade", "Burger", "Special Dessert",
        "Salad", "Pancake", "Coffee", "Tea"
    ],
    "Category": [
        "Drink", "Drink", "Drink", "Drink", "Food",
        "Food", "Drink", "Drink", "Drink", "Drink",
        "Food", "Food", "Food", "Drink", "Drink",
        "Drink", "Food", "Food", None, "Food",
        "Drink", "Drink"
    ],
    "Price": [
        120, 220, 80, 180, 450, 300, 100, 200, 250, 50,
        160, None, 500, 280, 210, 150, 650, 25000, 320, 380,
        120, 80
    ],
    "Rating": [
        4.5, 4.7, 4.2, 4.4, 4.1, None, 4.6, 4.5, 4.3, 4.0,
        4.2, 4.0, 4.3, 4.6, 4.5, 4.1, 4.2, 4.9, 4.0, 4.5,
        4.5, 4.2
    ],
    "Available": [
        "Yes", "Yes", "Yes", "Yes", "Yes", "Yes", "Yes", "Yes",
        "Yes", "Yes", "Yes", "Yes", "Yes", "Yes", None, "Yes",
        "Yes", "Yes", "Yes", "Yes", "Yes", "Yes"
    ],
    "Calories": [
        5, 150, 2, 110, 350, 420, 5, 120, 180, 0,
        280, 330, 400, 220, 190, 100, None, 500, 250, 450,
        5, 2
    ],
    "Size": [
        "Medium", "Large", "Small", "Medium", "Large",
        "Medium", "Small", "Medium", "Large", "Small",
        "Medium", "Medium", "Large", "Large", "Medium",
        None, "Large", "Large", "Medium", "Large",
        "Medium", "Small"
    ],
    "Supplier_Note": [
        "House special"
    ] + [None] * 21
})

raw_data = data.copy()

print("Shape:", data.shape)
data.head()
```

```python

rows_before = len(data)

data = data.drop_duplicates().reset_index(drop=True)

print("Duplicate rows removed:", rows_before - len(data))
data.shape
```

```python

missing_values = data.isnull().sum()

missing_table = pd.DataFrame({
    "Missing values": missing_values,
    "Percentage": missing_values / len(data) * 100
})

missing_table.sort_values(
    "Percentage", ascending=False
).round(2)
```

```python

missing_threshold = 70

columns_to_remove = missing_table[
    missing_table["Percentage"] >= missing_threshold
].index.tolist()

columns_to_remove
```

```python

data = data.drop(columns=columns_to_remove)

print("Columns removed:", columns_to_remove)
data.shape
```

```python

complete_rows = data.dropna()

salary_rows = data.dropna(subset=["Price"])

row_removal_comparison = pd.DataFrame({
    "Rows kept": [
        len(data), len(complete_rows), len(salary_rows)
    ],
    "Rows removed": [
        0,
        len(data) - len(complete_rows),
        len(data) - len(salary_rows)
    ]
}, index=[
    "Before row removal",
    "Remove every incomplete row",
    "Remove rows missing Price"
])

row_removal_comparison
```

```python

data.isnull().sum()
```

```python

numeric_columns = data.select_dtypes(include=np.number).columns
numeric_columns
```

```python

mean_example = data.copy()

for column in mean_example.select_dtypes(include=np.number).columns:
    mean_value = mean_example[column].mean()
    mean_example[column] = mean_example[column].fillna(mean_value)

imputation_comparison = pd.DataFrame({
    "Missing before": data[numeric_columns].isnull().sum(),
    "Mean": data[numeric_columns].mean(),
    "Median": data[numeric_columns].median(),
    "Missing after imputation":
        mean_example[numeric_columns].isnull().sum()
})

imputation_comparison
```

```python

numerical_missing_before = data[numeric_columns].isnull().sum()

for column in numeric_columns:
    data[column] = data[column].fillna(data[column].median())

pd.DataFrame({
    "Missing before": numerical_missing_before,
    "Missing after": data[numeric_columns].isnull().sum()
})
```

```python

categorical_columns = data.select_dtypes(include="object").columns

categorical_missing_before = data[categorical_columns].isnull().sum()

for column in categorical_columns:
    if data[column].isnull().any():
        data[column] = data[column].fillna(data[column].mode()[0])

pd.DataFrame({
    "Missing before": categorical_missing_before,
    "Missing after": data[categorical_columns].isnull().sum()
})
```

```python

outlier_columns = ["Price"]

Q1 = data[outlier_columns].quantile(0.25)
Q3 = data[outlier_columns].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

pd.DataFrame({
    "Q1": Q1,
    "Q3": Q3,
    "IQR": IQR,
    "Lower bound": lower_bound,
    "Upper bound": upper_bound
})
```

```python

outlier_count = {}

for column in outlier_columns:
    condition = (
        (data[column] < lower_bound[column]) |
        (data[column] > upper_bound[column])
    )
    outlier_count[column] = condition.sum()

pd.Series(outlier_count, name="Outliers")
```

```python

plt.figure(figsize=(8, 3))
plt.boxplot(data["Price"], vert=False)
plt.xlabel("Price")
plt.title("Cafe Menu Prices")
plt.show()
```

```python

price_q1 = data["Price"].quantile(0.25)
price_q3 = data["Price"].quantile(0.75)
price_iqr = price_q3 - price_q1

price_lower = price_q1 - 1.5 * price_iqr
price_upper = price_q3 + 1.5 * price_iqr

data_capped = data.copy()

data_capped["Price"] = data_capped["Price"].clip(
    lower=price_lower,
    upper=price_upper
)

outside_before = (
    (data["Price"] < price_lower) |
    (data["Price"] > price_upper)
).sum()

outside_after = (
    (data_capped["Price"] < price_lower) |
    (data_capped["Price"] > price_upper)
).sum()

print("Outliers before capping:", outside_before)
print("Outliers after capping:", outside_after)

data_capped["Price"].describe()
```

```python

numeric_columns = data_capped.select_dtypes(include=np.number).columns

standard_scaler = StandardScaler()

standardized_data = data_capped.copy()

standardized_data[numeric_columns] = standard_scaler.fit_transform(
    data_capped[numeric_columns]
)

standardized_data[numeric_columns].head()
```

```python

minmax_scaler = MinMaxScaler()

normalized_data = data_capped.copy()

normalized_data[numeric_columns] = minmax_scaler.fit_transform(
    data_capped[numeric_columns]
)

normalized_data[numeric_columns].head()
```

```python

label_encoder = LabelEncoder()

item_encoded = data_capped[["Item"]].copy()

item_encoded["Item_Code"] = label_encoder.fit_transform(
    item_encoded["Item"]
)

item_encoded.drop_duplicates().head(10)
```

```python

ordinal_encoder = OrdinalEncoder(
    categories=[["Small", "Medium", "Large"]]
)

size_encoded = data_capped[["Size"]].copy()

size_encoded["Size_Code"] = ordinal_encoder.fit_transform(
    size_encoded[["Size"]]
)

size_encoded.drop_duplicates().sort_values(
    "Size_Code"
).reset_index(drop=True)
```

```python

category_encoded = pd.get_dummies(
    data_capped["Category"],
    prefix="category",
    dtype=int
)

category_examples = pd.concat(
    [data_capped[["Category"]], category_encoded],
    axis=1
)

category_examples.drop_duplicates().reset_index(drop=True)
```

```python

encoded_data = pd.concat(
    [
        data_capped.drop(columns=["Category"]),
        category_encoded
    ],
    axis=1
)

encoded_data.head()
```
