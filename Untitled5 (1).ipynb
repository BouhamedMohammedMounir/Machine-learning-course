```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder, OrdinalEncoder

pd.set_option("display.max_columns", None)
```

```python
np.random.seed(42)

n = 1000

data = pd.DataFrame({
    "house_id": range(1, n + 1),
    "area": np.random.randint(50, 301, n),
    "bedrooms": np.random.randint(1, 7, n),
    "bathrooms": np.random.randint(1, 4, n),
    "floors": np.random.randint(1, 4, n),
    "year_built": np.random.randint(1980, 2026, n),
    "location": np.random.choice(
        ["El Oued", "Algiers", "Oran", "Constantine", "Biskra"],
        n
    ),
    "condition": np.random.choice(
        ["New", "Good", "Needs Renovation"],
        n
    ),
    "price": np.random.randint(3000000, 30000000, n)
})

raw_data = data.copy()

data.head()
```

```python
# Add missing values
data.loc[np.random.choice(data.index, 50), "area"] = np.nan
data.loc[np.random.choice(data.index, 40), "price"] = np.nan
data.loc[np.random.choice(data.index, 30), "location"] = np.nan
data.loc[np.random.choice(data.index, 30), "condition"] = np.nan

# Add duplicate rows
duplicates = data.sample(20, random_state=42)
data = pd.concat([data, duplicates], ignore_index=True)

# Add outliers
outlier_indices = data.sample(10, random_state=10).index
data.loc[outlier_indices, "price"] = 150000000

data.shape
```

```python
rows_before = len(data)

data = data.drop_duplicates().reset_index(drop=True)

print("Duplicate rows removed:", rows_before - len(data))

data.shape
```

```python
missing_values = data.isnull().sum()

missing_table = pd.DataFrame({
    "Missing values": missing_values,
    "Percentage": missing_values / len(data) * 100
})

missing_table.sort_values("Percentage", ascending=False).round(2)
```

```python
missing_threshold = 70

columns_to_remove = missing_table[
    missing_table["Percentage"] >= missing_threshold
].index.tolist()

columns_to_remove
```

```python
data = data.drop(columns=columns_to_remove)

print("Columns removed:", columns_to_remove)

data.shape
```

```python
complete_rows = data.dropna()

target_rows = data.dropna(subset=["price"])

row_removal_comparison = pd.DataFrame({
    "Rows kept": [
        len(data),
        len(complete_rows),
        len(target_rows)
    ],
    "Rows removed": [
        0,
        len(data) - len(complete_rows),
        len(data) - len(target_rows)
    ]
}, index=[
    "Before row removal",
    "Remove every incomplete row",
    "Remove rows missing price"
])

row_removal_comparison
```

```python
data.isnull().sum()
```

```python
numeric_columns = data.select_dtypes(include=np.number).columns

numeric_columns
```

```python
mean_example = data.copy()

for column in mean_example.select_dtypes(include=np.number).columns:
    mean_example[column] = mean_example[column].fillna(
        mean_example[column].mean()
    )

imputation_comparison = pd.DataFrame({
    "Missing before": data[numeric_columns].isnull().sum(),
    "Mean": data[numeric_columns].mean(),
    "Median": data[numeric_columns].median(),
    "Missing after imputation": mean_example[numeric_columns].isnull().sum()
})

imputation_comparison
```

```python
numerical_missing_before = data[numeric_columns].isnull().sum()

for column in numeric_columns:
    data[column] = data[column].fillna(data[column].median())

pd.DataFrame({
    "Missing before": numerical_missing_before,
    "Missing after": data[numeric_columns].isnull().sum()
})
```

```python
categorical_columns = data.select_dtypes(include="object").columns

categorical_missing_before = data[categorical_columns].isnull().sum()

for column in categorical_columns:
    if data[column].isnull().any():
        data[column] = data[column].fillna(data[column].mode()[0])

pd.DataFrame({
    "Missing before": categorical_missing_before,
    "Missing after": data[categorical_columns].isnull().sum()
})
```

```python
outlier_columns = ["price"]

Q1 = data[outlier_columns].quantile(0.25)
Q3 = data[outlier_columns].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

pd.DataFrame({
    "Q1": Q1,
    "Q3": Q3,
    "IQR": IQR,
    "Lower bound": lower_bound,
    "Upper bound": upper_bound
})
```

```python
outlier_count = {}

for column in outlier_columns:
    condition = (
        (data[column] < lower_bound[column]) |
        (data[column] > upper_bound[column])
    )
    outlier_count[column] = condition.sum()

pd.Series(outlier_count, name="Outliers")
```

```python
plt.figure(figsize=(8, 3))

plt.boxplot(data["price"], vert=False)

plt.xlabel("House Price")
plt.title("House Prices - Boxplot")

plt.show()
```

```python
price_q1 = data["price"].quantile(0.25)
price_q3 = data["price"].quantile(0.75)

price_iqr = price_q3 - price_q1

price_lower = price_q1 - 1.5 * price_iqr
price_upper = price_q3 + 1.5 * price_iqr

data_capped = data.copy()

data_capped["price"] = data_capped["price"].clip(
    lower=price_lower,
    upper=price_upper
)

outside_before = (
    (data["price"] < price_lower) |
    (data["price"] > price_upper)
).sum()

outside_after = (
    (data_capped["price"] < price_lower) |
    (data_capped["price"] > price_upper)
).sum()

print("Outliers before capping:", outside_before)
print("Outliers after capping:", outside_after)

data_capped["price"].describe()
```

```python
numeric_columns = data_capped.select_dtypes(include=np.number).columns

numeric_columns = numeric_columns.drop(
    ["house_id", "price"], errors="ignore"
)

standard_scaler = StandardScaler()

standardized_data = data_capped.copy()

standardized_data[numeric_columns] = standard_scaler.fit_transform(
    data_capped[numeric_columns]
)

standardized_data[numeric_columns].head()
```

```python
minmax_scaler = MinMaxScaler()

normalized_data = data_capped.copy()

normalized_data[numeric_columns] = minmax_scaler.fit_transform(
    data_capped[numeric_columns]
)

normalized_data[numeric_columns].head()
```

```python
label_encoder = LabelEncoder()

location_encoded = data_capped[["location"]].copy()

location_encoded["location_code"] = label_encoder.fit_transform(
    location_encoded["location"]
)

location_encoded.drop_duplicates().head(10)
```

```python
ordinal_encoder = OrdinalEncoder(
    categories=[["Needs Renovation", "Good", "New"]]
)

condition_encoded = data_capped[["condition"]].copy()

condition_encoded["condition_code"] = ordinal_encoder.fit_transform(
    condition_encoded[["condition"]]
)

condition_encoded.drop_duplicates().sort_values(
    "condition_code"
).reset_index(drop=True)
```

```python
location_encoded = pd.get_dummies(
    data_capped["location"],
    prefix="location",
    dtype=int
)

location_examples = pd.concat(
    [data_capped[["location"]], location_encoded],
    axis=1
)

location_examples = (
    location_examples.drop_duplicates()
    .sort_values("location")
    .reset_index(drop=True)
)

location_examples
```

```python
encoded_data = pd.concat(
    [
        data_capped.drop(columns=["location"]),
        location_encoded
    ],
    axis=1
)

encoded_data.head()
```

```python
print("Original data shape:", raw_data.shape)

print("Final data shape:", encoded_data.shape)

print("Remaining missing values:",
      int(encoded_data.isnull().sum().sum()))

print("Duplicated rows:",
      encoded_data.duplicated().sum())

encoded_data.head()
```

```python
final_data = data_capped.copy()

# Ordinal Encoding for condition
condition_mapping = {
    "Needs Renovation": 0,
    "Good": 1,
    "New": 2
}

final_data["condition"] = final_data["condition"].map(condition_mapping)

# One-Hot Encoding for location
final_data = pd.get_dummies(
    final_data,
    columns=["location"],
    prefix="location",
    dtype=int
)

final_data.head()
```

```python
print("Final shape:", final_data.shape)

print("Missing values:", final_data.isnull().sum().sum())

print("Duplicate rows:", final_data.duplicated().sum())

print("Data types:")
print(final_data.dtypes)
```

```python
final_data.to_csv("house_prices_cleaned.csv", index=False)

print("Dataset saved successfully!")
```
